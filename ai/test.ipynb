더미 값으로 생성합니다. 실제 키는 이 파일을 직접 열어서 채워주세요 (아래 셀은 파일이 이미 있으면 덮어쓰지 않습니다).

```python
import os

env_path = '/content/.env'
if not os.path.exists(env_path):
    with open(env_path, 'w') as f:
        f.write('AIHUB_API=REPLACE_WITH_YOUR_API_KEY\n')
    print(f"Created {env_path} with a dummy AIHUB_API value — edit it with your real key before running 3-3.")
else:
    print(f"{env_path} already exists — leaving it as is.")
```
